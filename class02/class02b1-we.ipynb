```python
import gensim.downloader as api
wv = api.load("word2vec-google-news-300")

```

```python
# representación word2vec
wv['dog']
```

```python
wv['cat']
```

```python
wv['house']
```

```python
from scipy.spatial.distance import cosine
print('la distancia entre perro y gato',cosine(wv['dog'],wv['cat']).round(2))
```

```python
print('la distancia entre perro y casa',cosine(wv['dog'],wv['house']).round(2))
```
