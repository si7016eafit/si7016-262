## 1. Crear servidor MCP v2

```python
server_code=r'''
from mcp.server import MCPServer
mcp=MCPServer("si7016-demo",instructions="Servidor docente de la clase 03.")
@mcp.tool()
def contar_palabras(texto:str)->int:
    """Cuenta palabras de un texto."""
    return len(texto.split())
@mcp.tool()
def info_curso()->str:
    """Información básica de SI7016."""
    return "SI7016 - NLP Aplicado - 2026-2"
if __name__=="__main__":
    mcp.run()
'''
open("mcp_server.py","w",encoding="utf-8").write(server_code)
print("mcp_server.py creado")
```

## 2. Cliente v2 por stdio

```python
import sys
from mcp import Client, StdioServerParameters
from mcp.client.stdio import stdio_client
server=StdioServerParameters(command=sys.executable,args=["mcp_server.py"])
async def demo():
    async with Client(stdio_client(server)) as client:
        print("Protocolo:",client.protocol_version)
        tools=await client.list_tools()
        for t in tools.tools:
            print(t.name,t.description,t.input_schema)
        r=await client.call_tool("contar_palabras",{"texto":"MCP desacopla agentes y herramientas externas"})
        print(r.content)
await demo()
```
